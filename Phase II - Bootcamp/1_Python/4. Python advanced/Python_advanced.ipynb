## Working with Datetime format

```python
df = pd.DataFrame({
        'date': ['2/10/2000', '1/11/2003', '9/12/2003'],
        'value': [3, 1, 5]
    })
df.info()
```

```python
df['date'] = pd.to_datetime(df['date'])

df.info()
```

- **Day first format**

```python
df = pd.DataFrame({
        'date': ['2/10/2000', '1/11/2003', '9/12/2003'],
        'value': [3, 1, 5]
    })
df['date'] = pd.to_datetime(df['date'], dayfirst = True)
df
```

- **Custom Format**

```python
df = pd.DataFrame({
        'date': ['2-10-2000 8:57:33', '1-11-2003 3:45:5', '9-12-2003 13:43:50'],
        'value': [3, 1, 5]
    })
df['date'] = pd.to_datetime(df['date'], format = '%d-%m-%Y %H:%M:%S')
df
```

- **Handle Parsing Error**

```python
df = pd.DataFrame({
        'date': ['2-10-2000 8:57:33', 'a-11-2003 3:45:5', '9-12-2003 13:43:50'],
        'value': [3, 1, 5]
    })
df['date'] = pd.to_datetime(df['date'], errors = 'coerce')
df
```

- **Assemble datetime from multiple columns**

```python
df = pd.DataFrame({
    'day': [1, 3, 12],
    'month': [3, 1, 5],
    'year': [2012, 2000, 1997]
    })
df['date'] = pd.to_datetime(df)
df
```

```python
df = pd.DataFrame({
    'name': ['Andy', 'Dony', 'Sam'],
    'DoB': ['2-10-1997', '29-02-2004', '9-12-2000']
})

df['DoB'] = pd.to_datetime(df['DoB'], dayfirst = True)
df['day'] = df['DoB'].dt.day
df['month'] = df['DoB'].dt.month
df['year'] = df['DoB'].dt.year
df['week_of_year'] = df['DoB'].dt.isocalendar().week
df['day_of_week'] = df['DoB'].dt.dayofweek
df['is_leap_year'] = df['DoB'].dt.is_leap_year
```

```python
df
```

```python
dw = {
    0: 'Monday',
    1: 'Tuesday',
    2: 'Wednesday',
    3: 'Thursday',
    4: 'Friday',
    5: 'Saturday',
    6: 'Sunday'
}
```

```python
df['day_of_week'] = df['DoB'].dt.dayofweek.map(dw)
df
```

```python
today = pd.to_datetime('today')
df['age'] = today.year - df['year'] 
df
```

```python
# Haven't had birthday
b_md = df['DoB'].apply(lambda x: (x.month, x.day))
no_birthday = b_md > (today.month, today.day) 
df['age'] = df['age'] - no_birthday
df
```

```python
import copy
import json
import avro
from avro.datafile import DataFileWriter, DataFileReader
from avro.io import DatumWriter, DatumReader

schema = {
    'name': 'avro.example.user',
    'type': 'record',
    'fields': [
        {'name': 'name', 'type': 'string'},
        {'name': 'age', 'type' : 'int'}
    ]
}

schema_parsed = avro.schema.Parse(json.dumps(schema))
```

```python
# Write data to avro file

with open('users.avro', 'wb') as f:
    writer = DataFileWriter(f, DatumWriter(), schema_parsed)
    writer.append({'name': 'Sam', 'age': 32})
    writer.append({'name': 'Sofia', 'age': 33})
    writer.close()
```

```python
# Read data from avro file

with open('users.avro', 'rb') as f:
    reader = DataFileReader(f, DatumReader())
    metadata = copy.deepcopy(reader.meta)
    parsed_from_file = json.loads(metadata['avro.schema'])
    users = [user for user in reader]
    reader.close()
```
